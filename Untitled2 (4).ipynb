# 1. Selection Bias

# (a)

The sample only consists of readers who wrote about problems in their marriages and families. It's a form of voluntary bias because the study was geared directly towards those families that already have problems within marriage and family. The missing side are those who don't have problems with marriage and families and people happy with families because only then can we accurately gauge the percent of people who really regret having children. The result from the sample overstates the percentage because those who volunteered more than likely had family problems. A better way is to get a random sample size of families who have had children and ask them the question so there's a more generalized population not geared towards one result.

# (b)

The bias that fits this sample is non response bias. The sample is missing about 170,000 people because they didn't go through with the interview since they had the least free time and were unlikely to sit through the 20 min. Those who picked up tend to have more free time. This matters because it doesn't accurately gauge the population taking account only 15% of the population. It overstates the truth for the population. A better way would to send follow ups, offer incentives, or make the interview shorter because that way people are more inclined to respond. This way you can gauge the actual percent and remove non response bias

# (c)

It's a form of bias of selection on the dependent variable. Those missing are those who are not at clinics, pharmacies, and hospitals. Those who go to these places are more likely than not to already have health conditions and are seeking assistance. We need to gauge those who aren't at these places to get a more accurate number for those healthy or are sick and at home too. It overstates the truth for the population. A better way would be to test people at other places too not just health institutions and going to homes is too much unless you want to go door to door so you'll get a more broad population. 

# (d)

This is a form of Berkson's Paradox. What's missing is all the drugs that weren't approved. It matters because we're only seeing severe side effect drugs that passed because they have benefits that outweigh the side effects. This doesn't take into account the drugs that have severe side effects but are less beneficial. This filter caused the relationship to be negative. It overstates the truth of the population. A better way to collect the data would be to take into account the ones that didn't pass with more side effects.

# (e)

This a form of Size Bias. Those missing are those who have a gym membership but didn't check in that often. It matters because those who check into the gym are counted more times than someone who goes once. It overstates the population. Take into account the average of all people who have a membership rather than just check ins.

# (f)

This is a form of Survivorship bias. Missing are some people who started the drug but stopped due to side effects. This skews the actual percentage because they never reach the 3 years of refills so they get filtered out. Understates the truth of the population. A better way to collect is tracking the drug use from their 1st prescription

# (a)

```python
fig, ax = plt.subplots(figsize=(6, 4))

ax.hist(births['births'], bins=40, edgecolor='black')


ax.set_xlabel("births")
ax.set_ylabel("number of days")
ax.set_title("births")
```

the histogram is bimodal one peak at around 7800 and one at around 13000 which is about twice the height of the other. Each bar represents a range of birth counts and the height is the number of days with that count. the median would probably lie in the 11000 range. The maximum is around 13000 and minimums around 6000, 10000, and 16000. the first smaller peak is around 7800 and the second bigger one around 13000. The smaller peak could represent weekends, while the other peak would just be normal birthing days plus scheduled surgeries. Maybe somedays have more staff than certain days like holidays vs non holidays too.

# (c)

```python
b_day = births.groupby('day_of_week')['births'].sum()
b_day
```

```python
total_births = births['births'].sum()
total_births
```

```python
prop_day = (b_day / total_births).reset_index()
prop_day.sum()
```

```python
prop_day
```

```python
avg_pday = 1 / 7
```

```python
fig, ax = plt.subplots(figsize=(6, 4))

ax.bar(prop_day['day_of_week'], prop_day['births'])
ax.axhline(avg_pday, color='red', linestyle='-', label='all births')

ax.set_xlabel('Day of the Week')
ax.set_ylabel('Proportion of Births')
ax.set_title('Proportion of Births per day of the Week')
```

The empirical proportions do not match the theoretical line. Monday through friday hoever over the line while saturday and sunday fall under. This could be because many hospitals and clinics choose to schedule births(like c-sections) during the week days so that may cause them to cluster more on monday through friday.
